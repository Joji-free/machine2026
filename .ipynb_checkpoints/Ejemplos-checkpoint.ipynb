```python
def promedio_lista(numeros):
    """Calcula el promedio de una lista y devuelve su ID"""
    promedio = sum(numeros) / len(numeros)
    return promedio, id(numeros)

# Ejemplo de uso
mi_lista = [10, 20, 30, 40, 50]
prom, id_lista = promedio_lista(mi_lista)
print(f"Promedio: {prom}")
print(f"ID de la lista: {id_lista}")
```

```python
def multiplicar_por_dos(lista):
    resultado = []
    for elemento in lista:
        resultado.append(elemento * elemento)
    return resultado

numeros =[2, 4]
print(multiplicar_por_dos(numeros))
```

```python
def divir(alan):
    resultado =[]
    for elemento in alan:
        resultado.append(elemento // 2)
    return resultado

print(divir([200, 300, 500]))
```

```python
numero_a_palabra = {6 : "seis", 7 : "siete", 8 : "ocho", 9 : "nueve", 10 : "diez"}

[numero if numero <= 5 else numero_a_palabra[numero] for numero in range(1,11)]
```

```python
# Definimos el diccionario
numero_a_palabra = {6: "seis", 7: "siete", 8: "ocho", 9: "nueve", 10: "diez"}

# Opción 1: Usando un bucle for tradicional
resultado = []
for numero in range(1, 11):
    if numero <= 5:
        resultado.append(numero)
    else:
        resultado.append(numero_a_palabra[numero])
print("Opción 1:", resultado)

# Opción 2: Usando una función
def convertir_numero(n):
    if n <= 5:
        return n
    else:
        return numero_a_palabra[n]

resultado2 = [convertir_numero(n) for n in range(1, 11)]
print("Opción 2:", resultado2)

# Opción 3: Usando map() con una función lambda
resultado3 = list(map(lambda n: n if n <= 5 else numero_a_palabra[n], range(1, 11)))
print("Opción 3:", resultado3)

# Opción 4: Usando un operador ternario en un bucle
resultado4 = []
for n in range(1, 11):
    resultado4.append(n if n <= 5 else numero_a_palabra[n])
print("Opción 4:", resultado4)
```
