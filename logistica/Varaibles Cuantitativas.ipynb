```python
import numpy as np
X = np.array([0.5, 23, 0.3, 4.5, 0.3, 0.5,
              -28, -50, 60, -100, -10, -11, 13, 19, 1, 9])
```

```python
from scipy import stats
media = np.mean(X)
mediana = np.median(X)
moda, _ = stats.mode(X)
print('media: ', media)
print('mediana: ', mediana)
print('moda :', moda)
```

```python
import numpy as np
import matplotlib.pyplot as plt

Q1 = np.percentile(X, 25)
Q3 = np.percentile(X, 75)
RIC = Q3 - Q1
lim_inf = Q1 - 1.5*RIC
lim_sup = Q3 + 1.5*RIC

print('RIC :', RIC)
print('limite inferior :', lim_inf)
print('limite superior :', lim_sup)
```

```python
plt.boxplot(X)
plt.grid()
plt.title("Diagrama de caja y bigotes")
plt.show()
```

```python
desvest = np.std(X)
print('Desviación típica: ', desvest)
print('Varianza: ', desvest**2)
```

```python
asimetria = stats.skew(X)
curtosis = stats.kurtosis(X, fisher=True)
print('Asimetría ', np.round(asimetria, 4))
print('Curtosis ', np.round(curtosis, 4))

if curtosis > 0:
    print('Distribución Leptocúrtica')
elif curtosis < 0:
    print('Distribución Platicúrtica')
else:
    print('Distribución Mesocúrtica')
```

```python
from sklearn import datasets
from pprint import pprint
iris = datasets.load_iris()
X = iris.data
y = iris.target

# Longitud del pétalo (columna 2): asimetría
asimetria_petalo = stats.skew(X[:, 2])
print('Asimetría longitud del pétalo:', np.round(asimetria_petalo, 4))
if asimetria_petalo > 0:
    print('Asimetría positiva (cola a la derecha)')
elif asimetria_petalo < 0:
    print('Asimetría negativa (cola a la izquierda)')

# Ancho del sépalo (columna 1): curtosis
curtosis_sepalo = stats.kurtosis(X[:, 1], fisher=True)
print('\nCurtosis ancho del sépalo:', np.round(curtosis_sepalo, 4))
if curtosis_sepalo > 0:
    print('Distribución Leptocúrtica')
elif curtosis_sepalo < 0:
    print('Distribución Platicúrtica')
else:
    print('Distribución Mesocúrtica')
```

```python
fig, ax = plt.subplots()
nombres_clases = iris.target_names
nombres_vars = iris.feature_names
colores = ['blue', 'orange', 'green']

for clase in range(3):
    X_clase = X[y == clase]
    medias = np.mean(X_clase, axis=0)
    desviaciones = np.std(X_clase, axis=0)
    ax.errorbar(
        x=range(4),
        y=medias,
        yerr=desviaciones,
        label=nombres_clases[clase],
        color=colores[clase],
        fmt='-o',
        capsize=5
    )

ax.set_xticks(range(4))
ax.set_xticklabels(nombres_vars, rotation=15)
ax.set_ylabel('Valor medio ± desviación típica')
ax.set_title('Media y desviación típica por clase')
ax.legend()
plt.tight_layout()
plt.show()
```
