```python
import numpy as np
from sklearn.impute import SimpleImputer
```

```python
X_train = [[1,      4], 
           [np.nan, 3], 
           [7,      8]]
```

```python
imputer = SimpleImputer(strategy='mean')
X_train_clean = imputer.fit_transform(X_train)
print(X_train_clean)
```

```python
X_test = [[np.nan, 10], 
          [6,      np.nan], 
          [8,      2]]

X_test_clean = imputer.transform(X_test)
print(X_test_clean)
```
