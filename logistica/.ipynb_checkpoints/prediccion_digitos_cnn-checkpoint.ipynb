# Prediccion de Digitos - CNN (mas precisa para imagenes propias)

```python
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Dense, Flatten, Dropout, BatchNormalization
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
from PIL import Image
import cv2

print('TensorFlow:', tf.__version__)
```

## Carga del dataset

```python
mnist = tf.keras.datasets.mnist
(x_train_full, y_train_full), (x_test, y_test) = mnist.load_data()

print('Entrenamiento:', x_train_full.shape)
print('Prueba:', x_test.shape)
```

## Preprocesamiento

```python
# Normalizar y agregar canal (requerido por Conv2D)
x_train_norm = x_train_full.reshape(-1, 28, 28, 1) / 255.0
x_test_norm  = x_test.reshape(-1, 28, 28, 1) / 255.0

# One-hot encoding
y_train_ohe = to_categorical(y_train_full, num_classes=10)
y_test_ohe  = to_categorical(y_test, num_classes=10)

# Division entrenamiento / validacion
x_tr, x_val, y_tr, y_val = train_test_split(
    x_train_norm, y_train_ohe,
    test_size=0.2,
    random_state=42
)

total = x_tr.shape[0] + x_val.shape[0] + x_test_norm.shape[0]
print(f'Entrenamiento : {x_tr.shape[0]} ({x_tr.shape[0]/total*100:.1f}%)')
print(f'Validacion    : {x_val.shape[0]} ({x_val.shape[0]/total*100:.1f}%)')
print(f'Prueba        : {x_test_norm.shape[0]} ({x_test_norm.shape[0]/total*100:.1f}%)')
```

## Data Augmentation
Genera variaciones de las imagenes de entrenamiento para que el modelo sea mas robusto ante imagenes del mundo real (rotaciones, desplazamientos, zoom).

```python
datagen = ImageDataGenerator(
    rotation_range=10,      # Rotar hasta 10 grados
    width_shift_range=0.1,  # Desplazar horizontalmente
    height_shift_range=0.1, # Desplazar verticalmente
    zoom_range=0.1          # Zoom leve
)
datagen.fit(x_tr)
print('Data augmentation configurado')
```

## Modelo CNN

```python
model = Sequential([
    # Bloque 1
    Conv2D(32, (3,3), activation='relu', padding='same', input_shape=(28,28,1)),
    BatchNormalization(),
    Conv2D(32, (3,3), activation='relu', padding='same'),
    MaxPooling2D(2,2),
    Dropout(0.25),

    # Bloque 2
    Conv2D(64, (3,3), activation='relu', padding='same'),
    BatchNormalization(),
    Conv2D(64, (3,3), activation='relu', padding='same'),
    MaxPooling2D(2,2),
    Dropout(0.25),

    # Clasificador
    Flatten(),
    Dense(256, activation='relu'),
    BatchNormalization(),
    Dropout(0.5),
    Dense(10, activation='softmax')
])

model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
model.summary()
```

## Entrenamiento

```python
early_stop = EarlyStopping(monitor='val_accuracy', patience=5, restore_best_weights=True)

H = model.fit(
    datagen.flow(x_tr, y_tr, batch_size=128),
    validation_data=(x_val, y_val),
    epochs=30,
    callbacks=[early_stop]
)
```

## Curvas de entrenamiento

```python
plt.style.use('ggplot')
plt.figure()
plt.plot(H.history['loss'],         label='train_loss')
plt.plot(H.history['val_loss'],     label='val_loss')
plt.plot(H.history['accuracy'],     label='train_acc')
plt.plot(H.history['val_accuracy'], label='val_acc')
plt.title('Training Loss and Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Loss / Accuracy')
plt.legend()
plt.show()
```

## Evaluacion en conjunto de prueba

```python
loss, acc = model.evaluate(x_test_norm, y_test_ohe, verbose=0)
print(f'Loss prueba     : {loss:.4f}')
print(f'Accuracy prueba : {acc*100:.2f}%')

predictions = model.predict(x_test_norm, verbose=0)
print(classification_report(y_test_ohe.argmax(axis=1), predictions.argmax(axis=1)))
```

## Prediccion con imagen propia

Coloca tu imagen en la misma carpeta y cambia el nombre en `RUTA_IMAGEN`.

Recomendaciones para mejores resultados:
- Fondo blanco, digito escrito en negro y grueso
- Que el numero ocupe la mayor parte de la imagen
- Formato PNG o JPG

```python
def preprocesar_imagen(ruta):
    # Cargar en escala de grises
    img = cv2.imread(ruta, cv2.IMREAD_GRAYSCALE)

    # Binarizacion con Otsu: separa fondo del digito automaticamente
    _, img_bin = cv2.threshold(img, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)

    # Encontrar el contorno del digito y recortarlo (elimina margen vacio)
    coords = cv2.findNonZero(img_bin)
    x, y, w, h = cv2.boundingRect(coords)
    img_crop = img_bin[y:y+h, x:x+w]

    # Agregar margen del 20% para que no quede pegado al borde (igual que MNIST)
    margin = int(max(w, h) * 0.2)
    img_pad = cv2.copyMakeBorder(img_crop, margin, margin, margin, margin,
                                  cv2.BORDER_CONSTANT, value=0)

    # Redimensionar a 28x28
    img_28 = cv2.resize(img_pad, (28, 28), interpolation=cv2.INTER_AREA)

    # Normalizar
    img_norm = img_28 / 255.0

    return img_norm


RUTA_IMAGEN = 'mi_numero.png'  # <-- cambia esto

img_proc = preprocesar_imagen(RUTA_IMAGEN)

# Mostrar imagen tal como la ve el modelo
plt.figure(figsize=(3, 3))
plt.imshow(img_proc, cmap='gray_r')
plt.title('Imagen preprocesada (28x28)')
plt.axis('off')
plt.show()

# Predecir
img_input = img_proc.reshape(1, 28, 28, 1)
probs  = model.predict(img_input, verbose=0)[0]
digito = probs.argmax()

print(f'Digito predicho : {digito}')
print(f'Confianza       : {probs[digito]*100:.2f}%')
print()
print('Probabilidades por digito:')
for i, p in enumerate(probs):
    marca = ' <-- prediccion' if i == digito else ''
    print(f'  {i}: {p*100:.2f}%{marca}')
```
