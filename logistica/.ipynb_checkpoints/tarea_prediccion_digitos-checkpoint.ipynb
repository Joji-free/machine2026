# 🔢 Predicción de Dígitos con Red Neuronal (MNIST)
### Tarea: Subir una imagen con un número y predecir qué dígito es
---

## 1. Importación de librerías

```python
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten, Dropout
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.optimizers import SGD
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
from PIL import Image, ImageOps
import warnings
warnings.filterwarnings('ignore')

print(f"TensorFlow versión: {tf.__version__}")
print("✅ Librerías cargadas correctamente")
```

## 2. Carga y preparación del dataset MNIST

```python
# Cargar el dataset MNIST
mnist = tf.keras.datasets.mnist
(x_train_full, y_train_full), (x_test, y_test) = mnist.load_data()

print("📊 Dimensiones originales del dataset:")
print(f"  Entrenamiento: {x_train_full.shape} imágenes, {y_train_full.shape} etiquetas")
print(f"  Prueba:        {x_test.shape} imágenes, {y_test.shape} etiquetas")
```

## 3. Visualización de muestras del dataset

```python
fig, axes = plt.subplots(2, 5, figsize=(12, 5))
fig.suptitle('Ejemplos del Dataset MNIST', fontsize=14, fontweight='bold')

for i, ax in enumerate(axes.flat):
    idx = np.random.randint(0, x_train_full.shape[0])
    ax.imshow(x_train_full[idx], cmap='gray_r')
    ax.set_title(f'Dígito: {y_train_full[idx]}', fontsize=10)
    ax.axis('off')

plt.tight_layout()
plt.show()
```

## 4. Preprocesamiento de datos
Se dividen los datos en **entrenamiento (72%)**, **validación (18%)** y **prueba (10%)**

```python
# Normalización: llevar pixeles al rango [0, 1]
x_train_norm = x_train_full / 255.0
x_test_norm  = x_test / 255.0

# One-hot encoding de las etiquetas
y_train_ohe = to_categorical(y_train_full, num_classes=10)
y_test_ohe  = to_categorical(y_test, num_classes=10)

# División en conjunto de entrenamiento y validación (90% / 10%)
x_tr, x_val, y_tr, y_val = train_test_split(
    x_train_norm, y_train_ohe,
    test_size=0.2,        # 20% validación del total de entrenamiento
    random_state=42
)

# Calcular porcentajes reales
total = x_tr.shape[0] + x_val.shape[0] + x_test_norm.shape[0]
pct_tr  = x_tr.shape[0] / total * 100
pct_val = x_val.shape[0] / total * 100
pct_te  = x_test_norm.shape[0] / total * 100

print("📐 División del dataset:")
print(f"  🟢 Entrenamiento : {x_tr.shape[0]:>6,} imágenes  ({pct_tr:.1f}%)")
print(f"  🟡 Validación    : {x_val.shape[0]:>6,} imágenes  ({pct_val:.1f}%)")
print(f"  🔴 Prueba        : {x_test_norm.shape[0]:>6,} imágenes  ({pct_te:.1f}%)")
print(f"  📦 Total         : {total:>6,} imágenes  (100.0%)")
```

## 5. Construcción de la Red Neuronal (MLP)

```python
# Arquitectura: Flatten → 512 neuronas ReLU → Dropout → 10 neuronas Softmax
model = Sequential([
    Flatten(input_shape=(28, 28)),          # Aplanar la imagen 28x28 → 784 valores
    Dense(512, activation='relu'),          # Capa oculta 1: 512 neuronas
    Dropout(0.3),                           # Regularización: desactiva 30% neuronas al azar
    Dense(256, activation='relu'),          # Capa oculta 2: 256 neuronas
    Dropout(0.2),                           # Regularización adicional
    Dense(10, activation='softmax')         # Capa de salida: 10 clases (0-9)
])

# Compilar el modelo
model.compile(
    loss='categorical_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)

print("🧠 Arquitectura del modelo:")
model.summary()
```

## 6. Entrenamiento del modelo

```python
EPOCHS = 15
BATCH_SIZE = 128

print(f"🏋️ Iniciando entrenamiento: {EPOCHS} épocas, batch size {BATCH_SIZE}")
print("-" * 60)

H = model.fit(
    x_tr, y_tr,
    validation_data=(x_val, y_val),
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    verbose=1
)

print("-" * 60)
print("✅ Entrenamiento completado")
```

## 7. Curvas de entrenamiento (Loss y Accuracy)

```python
epochs_range = np.arange(1, EPOCHS + 1)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle('Curvas de Entrenamiento', fontsize=14, fontweight='bold')

# Gráfica de Loss
ax1.plot(epochs_range, H.history['loss'],     label='Train Loss',     color='#e74c3c', linewidth=2)
ax1.plot(epochs_range, H.history['val_loss'], label='Val Loss',       color='#c0392b', linewidth=2, linestyle='--')
ax1.set_title('Pérdida (Loss)')
ax1.set_xlabel('Época')
ax1.set_ylabel('Loss')
ax1.legend()
ax1.grid(True, alpha=0.3)

# Gráfica de Accuracy
ax2.plot(epochs_range, H.history['accuracy'],     label='Train Accuracy', color='#2ecc71', linewidth=2)
ax2.plot(epochs_range, H.history['val_accuracy'], label='Val Accuracy',   color='#27ae60', linewidth=2, linestyle='--')
ax2.set_title('Precisión (Accuracy)')
ax2.set_xlabel('Época')
ax2.set_ylabel('Accuracy')
ax2.legend()
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('curvas_entrenamiento.png', dpi=150, bbox_inches='tight')
plt.show()

print(f"\n📈 Accuracy final entrenamiento : {H.history['accuracy'][-1]*100:.2f}%")
print(f"📈 Accuracy final validación    : {H.history['val_accuracy'][-1]*100:.2f}%")
```

## 8. Evaluación en conjunto de prueba

```python
print("🔍 Evaluando modelo en conjunto de PRUEBA...")
print("-" * 60)

# Evaluar métricas generales
loss_te, acc_te = model.evaluate(x_test_norm, y_test_ohe, verbose=0)
print(f"\n📊 Resultados en conjunto de PRUEBA:")
print(f"   Loss     : {loss_te:.4f}")
print(f"   Accuracy : {acc_te*100:.2f}%")

# Reporte detallado por clase
predictions = model.predict(x_test_norm, batch_size=128, verbose=0)
y_pred_cls  = predictions.argmax(axis=1)
y_true_cls  = y_test_ohe.argmax(axis=1)

print("\n📋 Reporte de clasificación por dígito:")
print("-" * 60)
print(classification_report(y_true_cls, y_pred_cls,
                             target_names=[f'Dígito {i}' for i in range(10)]))
```

## 9. Resumen de porcentajes de entrenamiento y prueba

```python
train_acc_final = H.history['accuracy'][-1] * 100
val_acc_final   = H.history['val_accuracy'][-1] * 100
test_acc_final  = acc_te * 100

print("=" * 50)
print("     RESUMEN DE RENDIMIENTO DEL MODELO")
print("=" * 50)
print(f"  🟢 Accuracy Entrenamiento : {train_acc_final:.2f}%")
print(f"  🟡 Accuracy Validación    : {val_acc_final:.2f}%")
print(f"  🔴 Accuracy Prueba        : {test_acc_final:.2f}%")
print("=" * 50)

# Gráfico de barras comparativo
fig, ax = plt.subplots(figsize=(8, 5))
categorias = ['Entrenamiento', 'Validación', 'Prueba']
valores    = [train_acc_final, val_acc_final, test_acc_final]
colores    = ['#2ecc71', '#f39c12', '#e74c3c']

bars = ax.bar(categorias, valores, color=colores, width=0.5, edgecolor='white', linewidth=1.5)

for bar, val in zip(bars, valores):
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.3,
            f'{val:.2f}%', ha='center', va='bottom', fontsize=12, fontweight='bold')

ax.set_ylim(90, 102)
ax.set_ylabel('Accuracy (%)', fontsize=12)
ax.set_title('Comparación de Accuracy por Subconjunto', fontsize=13, fontweight='bold')
ax.grid(axis='y', alpha=0.3)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.savefig('comparacion_accuracy.png', dpi=150, bbox_inches='tight')
plt.show()
```

## 10. 📸 Predicción con imagen propia
**Instrucciones:**
1. Guarda tu imagen en la misma carpeta que este notebook
2. Reemplaza `'mi_numero.png'` con el nombre de tu archivo
3. La imagen puede ser JPG o PNG, cualquier tamaño — se redimensiona automáticamente
4. Escribe el dígito en fondo blanco con trazo oscuro (o fondo negro con trazo blanco)

```python
def predecir_imagen(ruta_imagen):
    """
    Carga una imagen del disco, la preprocesa igual que MNIST
    y retorna la predicción del modelo con las probabilidades.
    """
    # 1. Cargar imagen y convertir a escala de grises
    img_original = Image.open(ruta_imagen).convert('L')

    # 2. Redimensionar a 28x28 (tamaño MNIST)
    img_28 = img_original.resize((28, 28), Image.LANCZOS)

    # 3. Convertir a array NumPy
    img_array = np.array(img_28)

    # 4. MNIST usa fondo negro y dígito blanco.
    #    Si la imagen tiene fondo claro, la invertimos.
    if img_array.mean() > 127:
        img_array = 255 - img_array

    # 5. Normalizar al rango [0, 1]
    img_norm = img_array / 255.0

    # 6. Añadir dimensión de batch: (1, 28, 28)
    img_input = np.expand_dims(img_norm, axis=0)

    # 7. Predecir con el modelo
    probs = model.predict(img_input, verbose=0)[0]
    digito_predicho = probs.argmax()
    confianza       = probs[digito_predicho] * 100

    # ─── Visualización ───────────────────────────────────────────────
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))
    fig.suptitle('Resultado de la Predicción', fontsize=14, fontweight='bold')

    # Imagen preprocesada
    ax1.imshow(img_norm, cmap='gray_r')
    ax1.set_title(f'Imagen ingresada (28×28)', fontsize=11)
    ax1.axis('off')

    # Barras de probabilidad
    colores_barras = ['#3498db'] * 10
    colores_barras[digito_predicho] = '#e74c3c'  # Resaltar predicción

    bars = ax2.bar(range(10), probs * 100, color=colores_barras,
                   edgecolor='white', linewidth=0.8)
    ax2.set_xlabel('Dígito', fontsize=11)
    ax2.set_ylabel('Probabilidad (%)', fontsize=11)
    ax2.set_title('Distribución de probabilidades', fontsize=11)
    ax2.set_xticks(range(10))
    ax2.grid(axis='y', alpha=0.3)
    ax2.spines['top'].set_visible(False)
    ax2.spines['right'].set_visible(False)

    # Anotar valor en barra ganadora
    ax2.text(digito_predicho, probs[digito_predicho]*100 + 1,
             f'{confianza:.1f}%', ha='center', fontsize=10,
             color='#e74c3c', fontweight='bold')

    plt.tight_layout()
    plt.show()

    # ─── Resultado final ─────────────────────────────────────────────
    print("\n" + "=" * 45)
    print(f"  🔢 DÍGITO PREDICHO  :  {digito_predicho}")
    print(f"  📊 CONFIANZA        :  {confianza:.2f}%")
    print("=" * 45)
    print("\n📋 Probabilidades por dígito:")
    for i, p in enumerate(probs):
        barra  = '█' * int(p * 30)
        marca  = ' ← PREDICCIÓN' if i == digito_predicho else ''
        print(f"  {i}: {barra:<30} {p*100:5.2f}%{marca}")

    return digito_predicho, confianza


# ══════════════════════════════════════════════════════════════════════
#  ▼▼▼  CAMBIA AQUÍ el nombre de tu imagen  ▼▼▼
RUTA_MI_IMAGEN = 'mi_numero.png'
# ══════════════════════════════════════════════════════════════════════

digito, confianza = predecir_imagen(RUTA_MI_IMAGEN)
```
