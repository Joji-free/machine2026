# Estandarización de Datos con StandardScaler

En este ejemplo aplicaremos estandarización a un conjunto de datos usando `StandardScaler` de `sklearn.preprocessing`.

```python
from sklearn import preprocessing
import numpy as np
```

## Carga de datos

Creamos el conjunto de datos de entrenamiento `X_train`.

```python
# Carga de datos
X_train = np.array([[ 1., -1.,  2.],
                    [ 2.,  0.,  0.],
                    [ 0.,  1., -1.]])
```

## Limpieza de datos: Estandarización

Utilizamos `StandardScaler` para centrar los datos en media 0 y desviación estándar 1.

```python
standardizer = preprocessing.StandardScaler()

# Estandarizamos los datos de entrenamiento
X_train_std = standardizer.fit_transform(X_train)

print(X_train_std)
```

## Copia del conjunto de datos

Creamos una segunda matriz para comprobar que `fit()` y `transform()` por separado producen el mismo resultado.

```python
X_train_2 = np.array([[ 1., -1.,  2.],
                      [ 2.,  0.,  0.],
                      [ 0.,  1., -1.]])
```

## Aplicación separada de `fit()` y `transform()`

Primero ajustamos el modelo con `fit()` y luego transformamos los datos con `transform()`.

```python
f = preprocessing.StandardScaler()
f = f.fit(X_train)

X_train_2_std = f.transform(X_train_2)

print(X_train_2_std)
```

## Verificación de resultados

Comprobamos si ambas matrices estandarizadas son iguales.

```python
print(np.array_equiv(X_train_std, X_train_2_std))
```
