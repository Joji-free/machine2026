```python
import numpy as np
from sklearn import datasets
from sklearn.dummy import DummyClassifier
from sklearn.model_selection import cross_val_predict, cross_validate, cross_val_score
from sklearn.model_selection import KFold
import sklearn.metrics as metrics
from sklearn.model_selection import train_test_split
from sklearn import preprocessing
from pprint import pprint
from sklearn.svm import SVC
```

```python
datos = datasets.load_iris()
print(np.shape(datos.data))
```

```python
X = datos.data
y = datos.target
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    datos.data, datos.target, test_size=0.2, random_state=42
)
print('Train dimensions: ', np.shape(X_train))
print('Test dimensions:  ', np.shape(X_test))
```

```python
standardizer = preprocessing.StandardScaler()
X_train = standardizer.fit_transform(X_train)
X_test = standardizer.transform(X_test)
```

```python
alg = SVC(C=0.1, gamma='auto', kernel='rbf')
```

```python
y_pred = cross_val_predict(alg, X_train, y_train,
                           cv=KFold(n_splits=5, shuffle=True, random_state=42))
pprint(y_pred)
```

```python
print("Exactitud: %.3f" % (metrics.accuracy_score(y_train, y_pred)))
print("Precisión: %.3f" % (metrics.precision_score(y_train, y_pred, average="macro")))
print("Sensibilidad: %.3f" % (metrics.recall_score(y_train, y_pred, average="macro")))
print("F1-score: %.3f" % (metrics.f1_score(y_train, y_pred, average="macro")))
```

```python
print("Matriz de confusión:\n", metrics.confusion_matrix(y_train, y_pred))
```

```python
print("Tabla de métricas:\n", metrics.classification_report(y_train, y_pred))
```

```python
model = SVC(C=0.1, gamma='auto', kernel='rbf')
model = model.fit(X_train, y_train)

test_results = model.score(X_test, y_test)
print('Exactitud en test: ', np.round(test_results * 100, 4), '%')
```

```python
y_pred_test = model.predict(X_test)
print(metrics.classification_report(y_test, y_pred_test))
print(metrics.confusion_matrix(y_test, y_pred_test))
```
